```python
#PARTE 1
produtos = ["Camiseta Básica", "Calça Jeans", "Jaqueta Corta-Vento"]
precos = [49.90, 129.90, 219.90]

for i, j in zip(produtos, precos):
  print(i,j)

  #PARTE 2
print("")
print("Adicionar produto")
peca = input("Digite o nome do produto")
preco = float(input("Digite o valor do produto"))

produtos.append(peca)
precos.append(preco)

#PARTE 3
print("")
len (produtos)
if len(produtos) > 0:
  produtos.pop(0)
  precos.pop(0)
else :
  print("Estoque vazio")

for i, j in zip(produtos, precos):
    print(f"{i} - R$ {j:.2f}")
```
